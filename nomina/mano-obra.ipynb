```python
#importar excel
import pandas as pd
path_informe="docs/manodeobra/1.Informe de Costo CTABARCA junio 2025.xlsx"
path_ods="docs/manodeobra/050/Empleados contrato N° 3023604 (junio - 2025).xlsx"
#leer dos archivos
informe_df = pd.read_excel(path_informe, sheet_name="Informe")
ods_df = pd.read_excel(path_ods)

#tomar como encabezado la fila [8]
informe_df.columns = informe_df.iloc[8]
informe_df = informe_df[9:]
print(f"Columnas del informe: {informe_df.columns.tolist()}")
# Tomar unicamente estas columnas:  ['Valor pagado en nomina ', 'ID', 'Identificación', 'Nro. Contrato', 'Nombres', 'Apellidos', 'Cargo', 'Fecha Inicio', 'Fecha Vencimiento', 'Nombre Centro Costo', 'estado', 'Fecha de Ingreso ', 'Fecha de retiro ', 'Días Trabajados']
informe_df = informe_df[['Valor pagado en nomina ', 'ID', 'Identificación', 'Nro. Contrato', 'Nombres', 'Apellidos', 'Cargo', 'Fecha Inicio', 'Fecha Vencimiento', 'Nombre Centro Costo', 'estado', 'Fecha de Ingreso ', 'Fecha de retiro ', 'Días Trabajados']]
informe_df = informe_df[informe_df['Valor pagado en nomina '] == "Recobro"]
```

```python
print(informe_df["Nombre Centro Costo"].unique())
#Extrar de "Nombre Centro Costo" el ODS de esta manera el valor al lado Os, ejemplo 'Ecop P&C Tabarca Grb Os054 - Operaciones-Ops (Cd)' ==> '54'
informe_df["ODS"] = informe_df["Nombre Centro Costo"].str.extract(r'Os0*(\d+)').astype(str)
print(informe_df["ODS"].unique())
```

```python
ods_df.columns.tolist()

#No_de_orden_de_servicio_conocido_por_el_contratista,'CargoContratoLaboral', NumeroDocumento, Nombres, Apellidos, 'Fecha_de_inicio_de_actividades_del_trabajador_para_el_contrato_comercial_u_orden_de_servicio',  'Fecha_fin_de_actividades_del_trabajador_para_el_contrato_comercial_u_orden_de_servicio', 'DiasLaborados', 'DiasTrabajadosEnMes', EstadoActualTrabajador
columnas = ['No_de_orden_de_servicio_conocido_por_el_contratista','CargoContratoLaboral', 'NumeroDocumento', 'Nombres', 'Apellidos', 'Fecha_de_inicio_de_actividades_del_trabajador_para_el_contrato_comercial_u_orden_de_servicio',  'Fecha_fin_de_actividades_del_trabajador_para_el_contrato_comercial_u_orden_de_servicio', 'DiasLaborados', 'DiasTrabajadosEnMes', 'EstadoActualTrabajador']
ods_df = ods_df[columnas]
print(ods_df.columns.tolist())
ods_df["No_de_orden_de_servicio_conocido_por_el_contratista"].to_string()
print(ods_df["No_de_orden_de_servicio_conocido_por_el_contratista"].unique())
#numero de documento sin puntos ni comas
ods_df["NumeroDocumento"] = ods_df["NumeroDocumento"].astype(str).str.replace(".", "").str.replace(",", "")
```
